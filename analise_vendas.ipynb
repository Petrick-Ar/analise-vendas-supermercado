```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('vendas_supermercado.csv')
print("Visualização inicial dos dados:")
print(df.head())

df['Receita_Total'] = df['Quantidade'] * df['Preco_Unitario']
print("\nDados com Receita Total adicionada:")
print(df.head())

receita_categoria = df.groupby('Categoria')['Receita_Total'].sum()
print("\nReceita total por categoria:")
print(receita_categoria)

plt.figure(figsize=(8, 5))
receita_categoria.plot(kind='bar', title='Receita por Categoria', color='skyblue')
plt.xlabel('Categoria')
plt.ylabel('Receita Total')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
