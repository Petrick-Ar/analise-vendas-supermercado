```python
# Análise de Vendas em Supermercado

# 1. Importando bibliotecas
import pandas as pd
import matplotlib.pyplot as plt

# 2. Carregando os dados (suba o arquivo vendas_supermercado.csv no Colab antes de rodar esta célula)
df = pd.read_csv('vendas_supermercado.csv')

# 3. Visualizando as primeiras linhas
print("Visualização inicial dos dados:")
print(df.head())

# 4. Criando uma nova coluna de Receita Total
df['Receita_Total'] = df['Quantidade'] * df['Preco_Unitario']
print("\nDados com Receita Total adicionada:")
print(df.head())

# 5. Agrupando a Receita por Categoria
receita_categoria = df.groupby('Categoria')['Receita_Total'].sum()
print("\nReceita total por categoria:")
print(receita_categoria)

# 6. Gerando gráfico de barras
plt.figure(figsize=(8, 5))
receita_categoria.plot(kind='bar', title='Receita por Categoria', color='skyblue')
plt.xlabel('Categoria')
plt.ylabel('Receita Total')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
